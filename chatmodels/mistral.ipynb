```python
from dotenv import load_dotenv
load_dotenv()
from langchain.chat_models import init_chat_model


model = init_chat_model("mistral-small-2603", temperature = 1)
```

temperature : 0 - 1 / 0-2

However, not every provider supports the full range. Some only allow 0.0 to 1.0.

1 : creative model

0: reasoning model/ mathematics / logic

| Temperature | Behavior                           | Best Use Cases                            |
| ----------- | ---------------------------------- | ----------------------------------------- |
| **0.0**     | Nearly deterministic, least random | Coding, math, factual Q&A, SQL generation |
| **0.2**     | Very focused                       | RAG, summarization, documentation         |
| **0.5**     | Balanced                           | General chatbot, explanations             |
| **0.7**     | More creative                      | Writing, brainstorming                    |
| **1.0**     | Highly creative                    | Stories, marketing copy                   |
| **1.5–2.0** | Very random and unpredictable      | Experimental creativity (rarely used)     |
